医療におけるリアルワールドデータ（RWD）は、日常診療などで収集される患者や医療に関するデータです。電子カルテ、レセプト、患者レジストリなどが代表例です。

- NDBは、厚生労働省が管理するレセプト情報や特定健診等のデータベースです。
- MID-NETは、協力医療機関の電子カルテやレセプト等を活用する、PMDAの医療情報データベースです。

## なぜRWDなのか？
日常診療の中で集められたデータであり、例えば
- 電子カルテ(EHR)
- DPCデータ
- レセプト
- 患者登録・レジストリ
- ウエアブル
- バイオマーカー
- 検査データ

臨床研究のゴールド・スタンダードはランダム化比較試験 （Randomized Controlled Trial: 以下 RCT）ですが、特に医療の世界では RCT が倫理的または費用的な面で実施困難であることが多く、また厳しく統制された RCT で得られた知見を実際の臨床に応用できるか保証されないという問題があるため（外的妥当性が保証されない）、近年 RWD を用いた分析の可能性に注目が集まっています。

1. 臨床試験の限界
従来の臨床試験（ランダム化比較試験、RCT）は、厳密な環境下で行われるため、患者の選定基準が非常に厳格です。これにより、現実の医療現場での患者層（高齢者、併存疾患を持つ患者など）を十分に反映できない場合があります。RWDは、日常の診療で得られる実際のデータを基にしているため、より多様な患者層に基づく実際の医療効果や安全性を評価できます。

2. 医薬品の効果・安全性の長期的評価
新薬が市場に出た後、臨床試験では確認できなかった長期的な安全性や副作用が現れることがあります。RWDは、医薬品が市場に出た後のリアルタイムでのデータを集め、長期間にわたる効果や副作用の監視に役立ちます。

3. 患者中心の医療へのシフト
医療は、個別化医療（Precision Medicine）や患者中心の医療に向かっています。RWDは、患者の日常生活や実際の医療現場での経験を反映するデータであり、個々の患者のニーズや治療効果をよりよく理解するための情報源となります。

4. コスト削減と効率化
従来の臨床試験は非常にコストが高く、時間もかかります。一方、電子カルテ（EHR）、保険データ、ウェアラブルデバイスなどから得られるRWDは、比較的低コストで迅速にデータを収集できるため、医薬品開発や政策決定において重要な役割を果たします。

5. 政策と規制の変化
2018年4月に施行された[改正GPSP省令](https://www.pmda.go.jp/files/000272301.pdf)で、医薬品の製造販売後データベース調査が新設されました。薬事承認においても、条件付き早期承認制度下での調査でRWDの活用が認められ、さらにはPMDA(独立行政法人医薬品医療機器総合機構)のレジストリ・データベースに関する相談窓口が整備されてきました。

6. デジタルヘルス技術の進化
ウェアラブルデバイス、スマートフォン、IoTなどのデジタル技術が普及し、個々の患者の日常生活での健康状態や行動をリアルタイムでモニタリングできるようになりました。これにより、RWDの質と量が飛躍的に向上し、より精密な分析が可能になっています。

このように、リアルワールドデータは従来の臨床試験で得られた知見を補うデータとして、医療・ヘルスケアの意思決定においてますます重要な役割を担っています。

RWDを用いた観察研究では、治療を受けた人と受けなかった人の背景の違い（交絡）や、欠測、記録方法の違いが結果に影響します。データの量に加え、研究目的に合ったデータかを確認します。

機械学習による予測モデルを評価するには、学習に使っていないデータに対して、予測と正解ラベルを比較します。

- **二値分類**は、陽性と陰性のように、データを二つのクラスに分ける課題です。
- **ロジスティック回帰**は、入力から陽性クラスの予測確率を求める手法です。その値をしきい値と比較して分類します。
- **深層学習**では、多層のニューラルネットワークを用いてデータの特徴を学習します。この演習では、画像分類に使われるVision Transformer（ViT）と畳み込みニューラルネットワーク（CNN）を比較します。

## **混同行列、ROC曲線、AUC**

```python
#こちらの動画を見てください。
from IPython.display import HTML

HTML("""
<a href="https://www.youtube.com/watch?v=opBZQOa8EZA" target="_blank">
▶ YouTubeで動画を見る
</a>
""")
```

混同行列では、行に真のラベル、列に予測ラベルを並べます。このノートでは、両方の軸をクラス0、クラス1の順に表示し、クラス1を陽性として扱います。

| 真のラベル ＼ 予測ラベル | クラス0（陰性） | クラス1（陽性） |
|---|---|---|
| クラス0（陰性） | TN（真陰性） | FP（偽陽性） |
| クラス1（陽性） | FN（偽陰性） | TP（真陽性） |

感度（TPR）は陽性のうち陽性と予測した割合、偽陽性率（FPR）は陰性のうち陽性と予測した割合です。特異度は $1-\mathrm{FPR}$ です。

しきい値を変えながら、横軸にFPR、縦軸にTPRを描いたものがROC曲線です。AUCはその曲線の下の面積です。無作為な順位づけの基準は0.5で、1に近いほど陽性に高いスコアを与える傾向が強いことを示します。正解率は一つのしきい値で測る指標です。

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'実行デバイス: {device}')
if device.type == 'cpu':
    print('CPUでも実行できますが、学習には時間がかかります。')
```

```python
# 両モデルへの入力を224×224のRGB画像にそろえます。
normalize = transforms.Normalize(
    mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225],
)
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    normalize,
])
eval_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    normalize,
])

train_data = datasets.ImageFolder(dataset_dir / 'train', train_transform)
valid_data = datasets.ImageFolder(dataset_dir / 'validation', eval_transform)
test_data = datasets.ImageFolder(dataset_dir / 'test', eval_transform)
if len(train_data.classes) != 2 or any(
    data.class_to_idx != train_data.class_to_idx for data in (valid_data, test_data)
):
    raise ValueError('訓練・検証・テストに、同じ名前のクラスフォルダが2つずつ必要です。')
print('クラス番号:', train_data.class_to_idx)
print(f'画像枚数  訓練: {len(train_data)}  検証: {len(valid_data)}  テスト: {len(test_data)}')

batch_size = 8
train_loader = DataLoader(train_data, batch_size=batch_size, shuffle=True)
valid_loader = DataLoader(valid_data, batch_size=batch_size, shuffle=False)
test_loader = DataLoader(test_data, batch_size=batch_size, shuffle=False)
```

この演習では100エポック学習します。検証損失が最も小さかった時点の重みを用いてテストデータを評価します。モデルの重みは動画と同様にマイドライブへ保存します。

## ViTで画像分類

```python
epochs = 100
learning_rate = 0.0001
seed = 42
```

```python
def train_model(model_name, checkpoint_path):
    torch.manual_seed(seed)
    model = timm.create_model(model_name, pretrained=True, num_classes=2).to(device)
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.AdamW(model.parameters(), lr=learning_rate)
    best_val_loss = float('inf')

    for epoch in range(epochs):
        train_loss, train_accuracy = run_epoch(model, train_loader, criterion, optimizer)
        val_loss, val_accuracy = run_epoch(model, valid_loader, criterion)
        print(
            f'{epoch + 1}/{epochs}  train loss={train_loss:.4f} '
            f'acc={train_accuracy:.4f}  val loss={val_loss:.4f} acc={val_accuracy:.4f}'
        )
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            torch.save(model.state_dict(), checkpoint_path)

    if not np.isfinite(best_val_loss):
        raise RuntimeError('検証損失が有限値にならず、重みを選べませんでした。')
    model.load_state_dict(torch.load(checkpoint_path, map_location=device, weights_only=True))
    return model


def run_epoch(model, loader, criterion, optimizer=None):
    training = optimizer is not None
    model.train(training)
    total_loss = 0.0
    total_correct = 0
    total_images = 0

    with torch.set_grad_enabled(training):
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)
            if training:
                optimizer.zero_grad()
            logits = model(images)
            loss = criterion(logits, labels)
            if training:
                loss.backward()
                optimizer.step()

            total_loss += loss.item() * labels.size(0)
            total_correct += (logits.argmax(dim=1) == labels).sum().item()
            total_images += labels.size(0)

    return total_loss / total_images, total_correct / total_images
```

```python
vit_path = drive_dir / 'bestViTmodel.pth'
model = train_model('vit_base_patch16_224.augreg_in21k', vit_path)
```

```python
# テストデータによる検証
def evaluate_model(model, test_loader):
    model.eval()
    predictions = []
    actuals = []
    probas = []
    with torch.no_grad():
        for images, labels in test_loader:
            images = images.to(device)
            labels = labels.to(device)
            outputs = model(images)
            predicted = outputs.argmax(dim=1)
            probas.extend(outputs.softmax(dim=1)[:, 1].cpu().numpy())  # 二番目のクラス（陽性クラス）の確率
            predictions.extend(predicted.cpu().numpy())
            actuals.extend(labels.cpu().numpy())


    # 指標の計算
    confusion = confusion_matrix(actuals, predictions, labels=[0, 1])
    accuracy = accuracy_score(actuals, predictions)
    recall = recall_score(actuals, predictions)  # 感度
    specificity = recall_score(actuals, predictions, pos_label=0)
    precision = precision_score(actuals, predictions, zero_division=np.nan)  # PPV
    npv = precision_score(actuals, predictions, pos_label=0, zero_division=np.nan)  # NPV
    f1 = f1_score(actuals, predictions, zero_division=np.nan)
    auc = roc_auc_score(actuals, probas)


    print(f"Accuracy: {accuracy:.4f}")
    print(f"感度: {recall:.4f}")
    print(f"特異度: {specificity:.4f}")
    print(f"PPV: {precision:.4f}")
    print(f"NPV: {npv:.4f}")
    print(f"F-Score: {f1:.4f}")
    print(f"AUC: {auc:.4f}")


    # 混同行列の可視化
    plt.figure(figsize=(7, 6))
    sns.heatmap(confusion, annot=True, fmt="d", cmap="Blues",
                xticklabels=test_data.classes, yticklabels=test_data.classes)
    plt.title("Confusion Matrix")
    plt.ylabel("True")
    plt.xlabel("Predicted")
    plt.show()


    # ROC曲線の描画
    fpr, tpr, _ = roc_curve(actuals, probas)
    plt.figure(figsize=(6,6))
    plt.plot(fpr, tpr, color='blue', label='ROC curve (AUC = %0.2f)' % auc)
    plt.plot([0, 1], [0, 1], color='gray', linestyle='--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('ROC Curve')
    plt.legend(loc="lower right")
    plt.show()
    return auc


vit_auc = evaluate_model(model, test_loader)
del model
```
